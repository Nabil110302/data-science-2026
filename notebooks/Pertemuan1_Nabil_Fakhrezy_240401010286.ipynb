## 1. Import Library dan Membuat Dataset


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

data = {
    "nama": ["Andi", "Budi", "Citra", "Dina", "Eka", "Fajar", "Gita", "Hadi"],
    "usia": [19, 20, 21, 19, 22, 20, 21, 23],
    "jam_belajar": [2, 3, 4, 2, 5, 3, 4, 6],
    "nilai": [70, 75, 82, 68, 88, 76, 84, 92],
}

df = pd.DataFrame(data)
display(df)
```

## 2. Statistik Deskriptif


```python
print("Shape:", df.shape)
display(df.describe().round(2))
```

## 3. Visualisasi Hubungan Jam Belajar dan Nilai


```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="jam_belajar", y="nilai", s=100)
sns.regplot(data=df, x="jam_belajar", y="nilai", scatter=False)
plt.title("Hubungan Jam Belajar dan Nilai")
plt.show()

display(df[["jam_belajar", "nilai"]].corr().round(3))
```
