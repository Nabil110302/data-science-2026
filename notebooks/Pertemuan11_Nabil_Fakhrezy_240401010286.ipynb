## Materi
K-Means, Hierarchical Clustering, Elbow Method, dan segmentasi pelanggan.


## 1. Import Library dan Generate Dataset Pelanggan


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import linkage, dendrogram

sns.set_theme(style="whitegrid")
np.random.seed(42)

n = 300
df = pd.DataFrame({
    "usia": np.random.randint(18, 60, n),
    "pendapatan_juta": np.random.normal(8, 3, n).clip(2, 20),
    "skor_belanja": np.random.randint(1, 101, n)
})

display(df.head())
display(df.describe().round(2))
```

## 2. Scaling dan Elbow Method


```python
fitur = ["usia", "pendapatan_juta", "skor_belanja"]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[fitur])

inertia = []
silhouette_scores = []
K = range(2, 11)

for k in K:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertia.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, labels))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(K, inertia, marker="o")
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("Jumlah Cluster")
axes[0].set_ylabel("Inertia")

axes[1].plot(K, silhouette_scores, marker="o")
axes[1].set_title("Silhouette Score")
axes[1].set_xlabel("Jumlah Cluster")
axes[1].set_ylabel("Score")
plt.show()
```

## 3. K-Means Clustering


```python
k_optimal = 3
kmeans = KMeans(n_clusters=k_optimal, random_state=42, n_init=10)
df["cluster_kmeans"] = kmeans.fit_predict(X_scaled)

display(df.groupby("cluster_kmeans")[fitur].mean().round(2))

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="pendapatan_juta", y="skor_belanja", hue="cluster_kmeans", palette="Set2")
plt.title("Segmentasi Pelanggan dengan K-Means")
plt.show()
```

## 4. Hierarchical Clustering dan Dendrogram


```python
linkage_matrix = linkage(X_scaled, method="ward")

plt.figure(figsize=(12, 5))
dendrogram(linkage_matrix, truncate_mode="lastp", p=20)
plt.title("Dendrogram Hierarchical Clustering")
plt.xlabel("Cluster")
plt.ylabel("Distance")
plt.show()

hier = AgglomerativeClustering(n_clusters=3, linkage="ward")
df["cluster_hierarchical"] = hier.fit_predict(X_scaled)

display(pd.crosstab(df["cluster_kmeans"], df["cluster_hierarchical"]))
```

Temuan utama adalah K-Means dapat membagi pelanggan berdasarkan pendapatan dan skor belanja, sedangkan dendrogram membantu melihat struktur pengelompokan. Keterbatasannya, data masih sintetis dan jumlah cluster tetap membutuhkan interpretasi bisnis.
